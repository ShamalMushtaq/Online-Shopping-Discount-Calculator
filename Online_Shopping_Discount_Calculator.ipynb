```python
# Online Shopping Discount Calculator

# for shopping amount
amount = float(input("Enter total shopping amount: Rs. "))

# for premium membership
premium = input("Are you a Premium Member? (Yes/No): ")

# Calculate regular discount using if-elif-else
if amount >= 10000:
    regular_discount_rate = 20
elif 5000 <= amount <= 9999:
    regular_discount_rate = 10
elif 2000 <= amount <= 4999:
    regular_discount_rate = 5
else:
    regular_discount_rate = 0

# Calculate regular discount amount
regular_discount = amount * regular_discount_rate / 100

# Nested if for Premium Member
premium_discount = 0

if premium.lower() == "yes":
    premium_discount = amount * 5 / 100

# Calculate total discount and final amount
total_discount = regular_discount + premium_discount
final_amount = amount - total_discount

# Display results
print("\n----- Shopping Bill -----")
print("Original Shopping Amount: Rs.", amount)
print("Regular Discount: Rs.", regular_discount)
print("Premium Discount: Rs.", premium_discount)
print("Total Discount: Rs.", total_discount)
print("Final Amount to Pay: Rs.", final_amount)
```
